```python
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime, timedelta
from matplotlib.patches import FancyBboxPatch
from matplotlib.font_manager import FontProperties
```

```python
def generate_dates(birth_date, current_date, num_years=90, weeks_per_year=52.1429):
    dates = []
    for year in range(num_years):
        for week in range(int(weeks_per_year)):
            week_date = birth_date + timedelta(weeks=year * weeks_per_year + week)
            dates.append((year, week, week_date <= current_date))
    return dates

def draw_life_calendar(birth_date_str, font_path='../fonts/Balsamiq_Sans/BalsamiqSans-Regular.ttf', dpi=100):
    '''
    Draw a life calendar from the birth date to the current date.
    The calendar is a grid where each cell represents a week of a year.
    The cells are colored in black if the week has already passed, and white otherwise.
    The calendar is displayed with the age on the y-axis and the week of the year on the x-axis.
    
    Parameters:
    - birth_date_str: string with the birth date in the format 'YYYY-MM-DD'
    - font_path: path to the font file to be used in the plot
    - dpi: dots per inch for the plot
    '''
    birth_date = datetime.strptime(birth_date_str, '%Y-%m-%d')
    current_date = datetime.now()
    dates = generate_dates(birth_date, current_date)

    fig, ax = plt.subplots(figsize=(10, 15))
    for year, week, is_past in dates:
        color = 'black' if is_past else 'white'
        # Creating a rectangle with rounded corners
        box = FancyBboxPatch((week, year), 0.9, 0.9, 
                             boxstyle="round,pad=-0.1,rounding_size=0.25",
                             edgecolor='gray',
                             facecolor=color,
                             linewidth=4.5)
        
        ax.add_patch(box)

    ax.set_xlim(-0.5, 51.5+0.9)  # Adjusted to align with the grid
    ax.set_ylim(-0.5, 89.5+0.9)  # Adjusted to align with the grid
    ax.set_aspect('equal')

    props = FontProperties(fname=font_path, size=16*4*2)
    ax.set_title('Life Calendar', pad=20*2*2, fontproperties=props)
    
    # to make x lable at the top
    ax.xaxis.set_label_position('top')
    props = FontProperties(fname=font_path, size=12*2*2)
    ax.set_xlabel('Week of the Year', fontproperties=props)
    # lable with added padding
    ax.xaxis.set_label_coords(0.5, 1.04)
    
    props = FontProperties(fname=font_path, size=12*2*2)
    ax.set_ylabel('Age', fontsize=12*2*2, fontproperties=props)
    # lable with added padding
    ax.yaxis.set_label_coords(-0.05, 0.5)

    # Setting ticks for years and weeks
    props = FontProperties(fname=font_path, size=10*2*2)
    ax.set_xticks(range(0, 52, 5))
    ax.set_xticklabels(range(0, 52, 5), fontproperties=props)
    ax.set_yticks(range(0, 91, 5))
    ax.set_yticklabels(range(0, 91, 5), fontproperties=props)
    
    # To make xticks at the top
    ax.xaxis.tick_top()

    ax.invert_yaxis()  # Invert the y-axis so that age increases as you go down
    
    plt.gcf().set_dpi(dpi)
    # set pixel size 
    plt.gcf().set_size_inches(23*2, 33*2)
    plt.show()
```

```python
BIRTH_DATE = '1990-01-01'  # Replace with your birth date

draw_life_calendar(
    birth_date_str=BIRTH_DATE,
    dpi=200)
```
